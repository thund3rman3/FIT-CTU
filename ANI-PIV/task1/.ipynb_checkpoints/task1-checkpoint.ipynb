```python
from pathlib import Path
import numpy as np
import cv2 as cv
import improutils as iu
import matplotlib.pyplot as plt

# Data is one level above this script folder: ANI-PIV/Data
image_dir = Path.cwd().resolve().parent / "Data"

if not image_dir.exists():
    raise NotADirectoryError(f"Folder not found: {image_dir}")

try:
    image_files = [p for p in image_dir.iterdir() if p.is_file()]
    #image_files = [image_files[0]]

except Exception as e:
    raise RuntimeError(f"Error accessing files in folder: {image_dir}") from e


images = []
for image_path in image_files:
    img = iu.load_image(image_path)
    if img is None:
        print(f"Skipping unreadable file: {image_path}")
        continue
    img = iu.to_gray(img)
    images.append(img)


#iu.show_images(*images.values(), scale=0.2)
#iu.plot_images(*images)
#plt.gcf().set_size_inches(6, 6)
iu.plot_images(images[0])
# print(f"Image shape: {images[0].shape}")
# iu.save_image(images[0], image_dir / "gray_image.png")
```

```python
import math


def plot_histograms(hists):
    n = len(hists)
    cols = 3
    rows = math.ceil(n / cols)

    fig, axes = plt.subplots(rows, cols, figsize=(12, 4*rows))

    axes = axes.flatten()

    for i, hist in enumerate(hists):
        axes[i].plot(hist)
        axes[i].set_title(f"Histogram {i}")
        axes[i].set_xlim([0, 256])

    plt.tight_layout()
    plt.show()
```

```python
from ipywidgets import interact, FloatSlider, IntSlider
def update_clahe(gray_img, clip_limit=2.0, tile_size=8):
    # CLAHE vyžaduje velikost mřížky alespoň (1, 1)
    tile_size = max(1, tile_size)
    
    clahe = cv.createCLAHE(clipLimit=clip_limit, tileGridSize=(tile_size, tile_size))
    return clahe.apply(gray_img)

# Vytvoření sliderů
interact(
    update_clahe,
    clip_limit=FloatSlider(min=0.5, max=40.0, step=0.5, value=2.0, description="Clip Limit:"),
    tile_size=IntSlider(min=2, max=64, step=2, value=8, description="Tile Size:")
);
```

```python
gamma = 2.5
# power_images = [
#     (cv.pow(img.astype(np.float32) / 255.0, gamma) * 255).astype(np.uint8)
#     for img in clahe_images
# ]
# iu.plot_images(*power_images)
power_images = [(cv.pow(images[0].astype(np.float32) / 255.0, gamma)
                * 255).astype(np.uint8)]
blurred = cv.medianBlur(images[0], 11)

cropped_pwr = power_images[0][300:600, 300:600]
cropped = images[0][300:600, 300:600]
cropped_blur = blurred[300:600, 300:600]
iu.plot_images(cropped, cropped_pwr, cropped_blur)
#iu.plot_images(blurred)
```

```python
# clipLimit
# Threshold for contrast limiting.
# Higher -> more contrast
# Lower -> less contrast

# tileGridSize
# Size of grid for histogram equalization. 
# Input image will be divided into equally sized rectangular tiles.
# Lower -> more details
# Higher -> less details

clahe = cv.createCLAHE(clipLimit=10.0, tileGridSize=(64, 64))
# clahe_images = [clahe.apply(img) for img in images]
# iu.plot_images(*clahe_images)
# clahe2 = cv.createCLAHE(clipLimit=30.0, tileGridSize=(8, 8))
# clahe_images2 = [clahe2.apply(img) for img in images]
# iu.plot_images(*clahe_images2)

clahe_images = [clahe.apply(blurred)]
iu.plot_images(images[0], clahe_images[0])
```

```python
#segmented_img = iu.segmentation_auto_threshold(power_images[0])
segmented_img = iu.segmentation_auto_threshold(clahe_images[0])
inverted_segmented_img_not = cv.bitwise_not(segmented_img)
# inverted_segmented_img = iu.erode(inverted_segmented_img_not, kernel_size=4, iterations=12,kernel=cv.MORPH_ELLIPSE)
# inverted_segmented_img = iu.dilate(inverted_segmented_img, kernel_size=5, iterations=5, kernel=cv.MORPH_ELLIPSE)

kernel_size = 5
iterations = 5
kernel = cv.getStructuringElement(cv.MORPH_ELLIPSE, (kernel_size, kernel_size))
inverted_segmented_img = cv.erode(inverted_segmented_img_not, kernel=kernel, iterations=iterations)
iterations = 5
inverted_segmented_img = cv.dilate(inverted_segmented_img, kernel=kernel, iterations=iterations)

iu.plot_images(
    clahe_images[0], inverted_segmented_img_not, inverted_segmented_img)

print(f": {inverted_segmented_img.shape}:{clahe_images[0].shape}")
iu.save_image(inverted_segmented_img, image_dir / "mask.png")
iu.save_image(clahe_images[0], image_dir / "img.png")
```

```python
def circularity(c):
    area = cv.contourArea(c)
    perimeter = cv.arcLength(c, True)
    if perimeter == 0:
        return 0
    return 4 * np.pi * area / (perimeter * perimeter)

def height_width_ratio(c):
    x, y, w, h = cv.boundingRect(c)
    if w == 0:
        return 0
    return h / w

min_area = 5111
max_area = 100000
external = True
fill = False
mode = cv.RETR_EXTERNAL
if not external:
    mode = cv.RETR_LIST
contours, _ = cv.findContours(
    inverted_segmented_img, mode, cv.CHAIN_APPROX_SIMPLE)
contours = [
    c
    for c in contours
    # and circularity(c) > 0.5
    if cv.contourArea(c) > min_area and cv.contourArea(c) < max_area and 1.35 > height_width_ratio(c) > 0.75 # cicularity(c) > 0.5
]  # 

contours_cnt = len(contours)

thick = cv.FILLED
if not fill:
    thick = 5
# contours_img = cv.drawContours(
#     np.zeros(inverted_segmented_img.shape, dtype=np.uint8),
#     contours,
#     -1,
#     color=(255, 255, 255),
#     thickness=thick,
# )
input_img = cv.cvtColor(inverted_segmented_img, cv.COLOR_GRAY2BGR)
contours_img = cv.drawContours(
    np.zeros(input_img.shape, dtype=np.uint8),
    contours,
    -1,           # všechny kontury
    color=(0, 0, 255),  # červená v BGR
    thickness=thick,
)


print(f"Number of contours found: {contours_cnt}")
print(contours_img.shape)
```

```python
img_RGB = cv.cvtColor(images[0], cv.COLOR_GRAY2BGR)
img_with_contours = cv.bitwise_or(img_RGB, contours_img)
iu.plot_images(contours_img, img_with_contours, inverted_segmented_img)
```
