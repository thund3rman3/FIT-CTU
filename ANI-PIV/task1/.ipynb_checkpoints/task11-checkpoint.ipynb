```python
from pathlib import Path
import numpy as np
import cv2 as cv
import improutils as iu
import matplotlib.pyplot as plt

# Data is one level above this script folder: ANI-PIV/Data
image_dir = Path.cwd().resolve().parent / "Data"
inverted_segmented_img = iu.load_image(image_dir / "mask.png")
inverted_segmented_img = iu.to_gray(inverted_segmented_img)
img = iu.load_image(image_dir / "PCD1.tif")
preproc_img = iu.load_image(image_dir/"img.png")
#img = iu.to_gray(img)
print(f": {inverted_segmented_img.shape}:{img.shape}")
```

```python
def circularity(c):
    area = cv.contourArea(c)
    perimeter = cv.arcLength(c, True)
    if perimeter == 0:
        return 0
    return 4 * np.pi * area / (perimeter * perimeter)


def height_width_ratio(c):
    x, y, w, h = cv.boundingRect(c)
    if w == 0:
        return 0
    return h / w


min_area = 5111
max_area = 100000
external = True
fill = False
mode = cv.RETR_EXTERNAL
if not external:
    mode = cv.RETR_LIST
contours, _ = cv.findContours(
    inverted_segmented_img, mode, cv.CHAIN_APPROX_SIMPLE)
contours = [
    c
    for c in contours
    # and circularity(c) > 0.5
    # cicularity(c) > 0.5
    if cv.contourArea(c) > min_area and cv.contourArea(c) < max_area and 1.55 > height_width_ratio(c) > 0.45 and circularity(c) > 0.6
    
]  

contours_cnt = len(contours)

thick = cv.FILLED
if not fill:
    thick = 5
# contours_img = cv.drawContours(
#     np.zeros(inverted_segmented_img.shape, dtype=np.uint8),
#     contours,
#     -1,
#     color=(255, 255, 255),
#     thickness=thick,
# )
input_img = cv.cvtColor(inverted_segmented_img, cv.COLOR_GRAY2BGR)
# contours_img = cv.drawContours(
#     np.zeros(input_img.shape, dtype=np.uint8),
#     contours,
#     -1,           # všechny kontury
#     color=(0, 0, 255),  # červená v BGR
#     thickness=thick,
# )
img_with_boxes = img.copy()
height, width = img.shape[:2]

x_max = width - 1
y_max = height - 1

img_grey = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
half_crop_x = half_crop_y = 85
cells=[]
masks = []

for contour in contours:
    x, y, w, h = cv.boundingRect(contour)
    center_x = x+w//2
    center_y = y+h//2
    if center_x-half_crop_x < 0 or center_y-half_crop_y < 0 or center_x+half_crop_x > x_max or center_y+half_crop_y > y_max:
        continue
        
    cv.rectangle(
        img_with_boxes,
        (center_x-half_crop_x, center_y-half_crop_y),
        (center_x+half_crop_x, center_y+half_crop_y),
        color=(0, 0, 255),  # červená v BGR
        thickness=thick,
    )
    masks.append(inverted_segmented_img[center_y - half_crop_y: center_y + half_crop_y, center_x -
                 half_crop_x: center_x + half_crop_x])
    cells.append(img_grey[center_y - half_crop_y: center_y + half_crop_y, center_x -
                 half_crop_x: center_x + half_crop_x])

print(f"Number of contours found: {contours_cnt}")
print(img_with_boxes.shape)
```

```python
from collections import deque

class Pixel:
   def __init__(self, row, col):
       self.row = row
       self.col = col


def image_reconstruct(marker: np.ndarray, mask: np.ndarray) -> np.ndarray:
   """
   Perform morphological reconstruction by dilation.

   Parameters:
       marker (np.ndarray): The marker image (seed).
       mask (np.ndarray): The mask image (constraint).

   Returns:
       np.ndarray: Reconstructed image.
   """
   if marker.shape != mask.shape:
       raise ValueError("Marker and mask must have the same dimensions")

   # Ensure images are in the same type
   marker = marker.astype(mask.dtype)

   # Create a working copy
   reconstructed = marker.copy()

   # Use a queue for pixels to update
   q = deque()
   rows, cols = reconstructed.shape

   # Initialize queue with pixels where marker < mask
   for r in range(rows):
       for c in range(cols):
           if reconstructed[r, c] < mask[r, c]:
               q.append(Pixel(r, c))

   # Neighbor offsets (4-connectivity)
   neighbors = [(-1, 0), (1, 0), (0, -1), (0, 1)]

   while q:
       p = q.popleft()
       for dr, dc in neighbors:
           rr, cc = p.row + dr, p.col + dc
           if 0 <= rr < rows and 0 <= cc < cols:
               new_val = min(mask[rr, cc], max(reconstructed[rr, cc], reconstructed[p.row, p.col]))
               if new_val > reconstructed[rr, cc]:
                   reconstructed[rr, cc] = new_val
                   q.append(Pixel(rr, cc))

   return reconstructed
```

```python
masked = masks[11]
kernel_size = 7
kernel = cv.getStructuringElement(cv.MORPH_ELLIPSE, (kernel_size, kernel_size))
#kernel = np.ones((kernel_size, kernel_size), np.uint8)
iterations = 8
eroded_img = cv.erode(masked, kernel=kernel, iterations=iterations)
masked = cv.dilate(eroded_img, kernel=kernel, iterations=iterations+5)


inverted_img = cv.bitwise_not(cells[11])
kernel_size = 3
#kernel = cv.getStructuringElement(cv.MORPH_ELLIPSE, (kernel_size, kernel_size))
kernel = np.ones((kernel_size, kernel_size), np.uint8)
iterations = 7
eroded_img = cv.erode(inverted_img, kernel=kernel, iterations=iterations)


recons = image_reconstruct(eroded_img,inverted_img )

iu.plot_images(cells[11],inverted_img)
iu.plot_images(eroded_img, recons)
```

```python
out = inverted_img - recons
out = iu.segmentation_auto_threshold(out)
out_mask = cv.bitwise_and(out, masked)
kernel_size = 2
kernel = np.ones((kernel_size, kernel_size), np.uint8)
iterations = 5
d = cv.erode(out_mask, kernel=kernel, iterations=iterations)
dilated = cv.dilate(d, kernel=kernel, iterations=iterations)
fin = cv.bitwise_or(dilated, cells[11])

iu.plot_images(masked, out, out_mask, d )
iu.plot_images(dilated, fin)
```
